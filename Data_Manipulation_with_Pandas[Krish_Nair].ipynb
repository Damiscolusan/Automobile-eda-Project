## Merging and Joining DataFrames


```python
df1 = pd.DataFrame({'Key':['A', 'B', 'C'], 'Value1':[1,2,3]})
df2 = pd.DataFrame({'Key':['A', 'B', 'D'], 'Value1':[4,5,6]})
```

```python
display('df1', df1)
print()
display('df2', df2)
```

Merge DataFrames on 'Key Columns'

```python
#Inner join is like intersection
# inner means from left to right at Matching key columns
pd.merge(df1, df2, on='Key', how='inner')
```

```python
#Outer join is like Union
pd.merge(df1, df2, on='Key', how='outer')
```

```python
# For the Left join the left DataFrame is given more priority than the right DataFrame
pd.merge(df1, df2, on='Key', how='left')# like left outer join
```

```python
# For the Right join the left DataFrame is given more priority than the left DataFrame
pd.merge(df1, df2, on='Key', how='right')# like right outer join
```
